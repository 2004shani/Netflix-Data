```python
# Duplicated Record
df.duplicated().sum()
```

```python
# Remove Duplicate
df.drop_duplicates(inplace=True)
```

```python
# Dirty Values- N/A, Null, Space, Blank
df.replace(
    ['N/A', 'Null','', ' ', '  '], np.nan, inplace=True
)
```

```python
df.isnull().sum()
```

```python
#Extra Space Remove
df= df.apply( lambda x: x.str.strip() if x.dtype== 'str' else x)
```

```python
# Standardize Text
df['Country'] =df['Country'].str.title()
df['Genre'] =df['Genre'].str.title()
df['Subscription_Plan'] =df['Subscription_Plan'].str.title()
df['Account_Status'] =df['Account_Status'].str.title()
```

```python
df['Country'].value_counts()
```

```python
df.columns
```

```python
# Numeric Columns
numeric_cols =  ['Age', 'Watch_Hours', 'Monthly_Fee', 'Movies_Watched', 'Episodes_Watched', 'Avg_Rating',
       'Monthly_Sessions', 'Download_Count']
```

```python
for col in numeric_cols:
    df[col]= pd.to_numeric(df[col], errors= 'coerce')
```

```python
df.describe()
```

```python
#Negative Values
df= df[(df['Age']>=18) & (df['Age']<=100)]
```

```python
df['Age'].describe()
```

```python
# Negative Values
df= df[df['Watch_Hours']>=0]
df= df[df['Monthly_Fee']>=0]
```

```python
df.describe()
```

```python
# Date Cleaning
df['Join_Date'] = pd.to_datetime(
    df['Join_Date'], 
    errors= 'coerce'
)

df['Last_Watch_Date'] = pd.to_datetime(
    df['Last_Watch_Date'], 
    errors= 'coerce'
)
```

```python
df[['Join_Date', 'Last_Watch_Date']].isnull().sum()
```

```python
# Missing Value Handle
df['Age']=df['Age'].fillna(
    df['Age'].mean()
)

df['Watch_Hours']=df['Watch_Hours'].fillna(
    df['Watch_Hours'].mean()
)

df['Monthly_Fee']=df['Monthly_Fee'].fillna(
    df['Monthly_Fee'].mean()
)
```

```python
df['Payment_Method']= df['Payment_Method'].fillna(
    "Unknow"
)

df['Device']=df['Device'].fillna(
    "Unknow"
)

df['Genre']=df['Genre'].fillna(
    "Unknow"
)
```

```python
# Feature Engineering
# Customer Revenue

df['Customer_Revenue'] = (df['Monthly_Fee'] * df['Monthly_Sessions'])
```

```python
df.head()
```

```python
# Engagement Score
df['Engagement_Score']= (df['Watch_Hours']+df['Monthly_Sessions']+df['Download_Count'])
```

```python
df.head()
```

```python
# Watch Category

bins= [0,20,50,100,np.inf]
label= ['Low', 'Medium', 'High', 'Very High']

df['Watch_Category']= pd.cut(df['Watch_Hours'], bins=bins, labels=label)
```

```python
df.head()
```

```python
df['Watch_Category'].value_counts()
```

```python
# Customer Age Group

bins= [17,25,35,50,100]
label= ['18-25', '26-35', '36-50', '51+']

df['Age_Group']= pd.cut(df['Age'], bins=bins, labels=label)
```

```python
df['Age_Group'].value_counts()
```

```python
# Account Active Flag

df['Active_Flag'] = np.where(df['Account_Status']== 'Active', 1, 0)
```
